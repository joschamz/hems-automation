```python
import pandas as pd
```

```python
df = pd.read_csv("data/household_data_15min_singleindex.csv")
df
```

```python
timestamp_col = "cet_cest_timestamp" if "cet_cest_timestamp" in df.columns else "utc_timestamp"
resident6_cols = [c for c in df.columns if "residential6" in c.lower()]

df_resident6 = df[[timestamp_col] + resident6_cols]
df_resident6.describe()
```

```python
df_em = df_resident6[(df_resident6[timestamp_col] >= "2016-07-10T00:00:00+0200") & (df_resident6[timestamp_col] < "2016-07-11T00:00:00+0200")]
df_em
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 5))
plt.plot(df_em[timestamp_col], df_em.DE_KN_residential6_grid_import, label="DE_KN_residential6_grid_import")
plt.xlabel("Time")
plt.ylabel("Cumulative Energy (kWh)")
plt.title("Energy Consumption of Residential6 on 2016-07-10")
plt.legend()
plt.show()  
```
